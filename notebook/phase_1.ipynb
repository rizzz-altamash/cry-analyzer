```python
import os
import librosa
import numpy as np
import noisereduce as nr
import soundfile as sf
```

```python
print("Normal:", len(os.listdir("dataset/normal")))
print("Sepsis:", len(os.listdir("dataset/sepsis")))
```

```python
# Preprocessing to normalize duration and suppress noise

TARGET_LEN = 16000
SR = 16000

def fix_length(y):
    if len(y) > TARGET_LEN:
        return y[:TARGET_LEN]
    else:
        return np.pad(y, (0, TARGET_LEN - len(y)))

def process_and_save(input_folder, output_folder):
    os.makedirs(output_folder, exist_ok=True)

    for file in os.listdir(input_folder):
        if file.endswith(".wav"):
            path = os.path.join(input_folder, file)

            y, sr = librosa.load(path, sr=SR)
            y = fix_length(y)
            y_clean = nr.reduce_noise(y=y, sr=sr)

            out_path = os.path.join(output_folder, file)
            sf.write(out_path, y_clean, sr)

# Run for both classes
process_and_save("dataset/normal", "clean/normal")
process_and_save("dataset/sepsis", "clean/sepsis")
```
